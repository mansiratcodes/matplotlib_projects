```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.patches import Ellipse
from matplotlib.colors import LinearSegmentedColormap
```

```python
#figure and dark background
fig,ax=plt.subplots(figsize=(9,9))
fig.patch.set_facecolor("#050816")
ax.set_facecolor("#050816")
#space boundaries
ax.set_xlim(-10,10)
ax.set_ylim(-10,10)
ax.set_aspect("equal")
ax.axis("off")

# Create a smooth space gradient
gradient_colors = ["#030611", "#10152F", "#30205A"]

space_cmap = LinearSegmentedColormap.from_list(
    "space_gradient",
    gradient_colors
)

# Generate the background grid
x = np.linspace(-10, 10, 500)
y = np.linspace(-10, 10, 500)
X, Y = np.meshgrid(x, y)

# Create a soft glow around the Sun
R = np.sqrt(X**2 + Y**2)
glow = np.exp(-(R / 5)**2)

# Draw the gradient behind everything
ax.imshow(
    glow,
    extent=[-10, 10, -10, 10],
    origin="lower",
    cmap=space_cmap,
    aspect="auto",
    zorder=0
)
#random stars
np.random.seed(42)
star_x=np.random.uniform(-10,10,150)
star_y=np.random.uniform(-10,10,150)
ax.scatter(
    star_x,star_y,
    s=3,color="white",alpha=0.7
)
# Sun glow layers — draw these before the Sun
glow_sizes = [2.0, 1.5, 1.1]
glow_alphas = [0.035, 0.06, 0.10]

for size, alpha in zip(glow_sizes, glow_alphas):
    glow_circle = Circle(
        (0, 0),
        size,
        color="#FFD43B",
        alpha=alpha,
        zorder=3
    )
    ax.add_patch(glow_circle)
#sun
sun=Circle(
    (0,0),0.65,
    color="#FFD43B",
    zorder=5
)
ax.add_patch(sun)
#planet orbits
orbit_radii=[1.5,2.2,3.0,4.0,5.2,6.5,7.8,9.0]
for radius in orbit_radii:
    orbit=Circle(
        (0,0),radius,
        fill=False,
        edgecolor="#34415F",
        linewidth=0.7,
        alpha=0.8
    )
    ax.add_patch(orbit)
ax.set_title(
    "SOLAR SYSTEM",
    color="white",
    fontsize=20,
    fontweight="bold",
    pad=20
)
# Planet properties
planet_names = [
    "Mercury", "Venus", "Earth", "Mars",
    "Jupiter", "Saturn", "Uranus", "Neptune"
]

planet_colors = [
    "#A6A6A6", "#E6B566", "#3B82F6", "#EF553B",
    "#D9B38C", "#E5C07B", "#5DE2E7", "#4169E1"
]

planet_sizes = [0.10, 0.17, 0.19, 0.14, 0.38, 0.32, 0.25, 0.24]

# Starting angles for the planets
angles = np.linspace(0, 2 * np.pi, 8, endpoint=False)

planets = []

for name, color, size, radius, angle in zip(
    planet_names, planet_colors, planet_sizes,
    orbit_radii, angles
):
    x = radius * np.cos(angle)
    y = radius * np.sin(angle)

    planet = Circle(
        (x, y),
        size,
        color=color,
        zorder=6
    )

    ax.add_patch(planet)
    planets.append(planet)


plt.show()
```

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Save the starting angles
initial_angles = angles.copy()

# Inner planets move faster than outer planets
speeds = [0.080, 0.060, 0.045, 0.035,
          0.025, 0.020, 0.015, 0.012]

def update(frame):
    for i, planet in enumerate(planets):
        angle = initial_angles[i] + frame * speeds[i]

        x = orbit_radii[i] * np.cos(angle)
        y = orbit_radii[i] * np.sin(angle)

        planet.center = (x, y)

    return planets

ani = FuncAnimation(
    fig,
    update,
    frames=300,
    interval=30,
    blit=True
)

plt.close(fig)
HTML(ani.to_jshtml())
```
