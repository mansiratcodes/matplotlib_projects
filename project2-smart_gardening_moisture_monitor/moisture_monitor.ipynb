```python
import matplotlib.pyplot as plt
```

```python
days=[1,2,3,4,5,6,7]
plant_a=[72,68,55,48,70,65,64]
plant_b=[80,82,89,90,85,91,89]
plant_c=[65,62,58,52,45,68,60]
```

```python
plt.plot(days,plant_a,label="Plant A",marker="o",lw=2)
plt.plot(days,plant_b,label="Plant B",marker="o",lw=2)
plt.plot(days,plant_c,label="Plant C",marker="o",lw=2)
print(min(plant_a))
print(min(plant_b))
print(min(plant_c))
print(plant_a.index(min(plant_a))+1)
print(plant_b.index(min(plant_b))+1)
print(plant_c.index(min(plant_c))+1)
plt.annotate("Driest",xy=(5,45),xytext=(5,55),arrowprops=dict(arrowstyle="->"))
lowest_moisture=min(min(plant_a),min(plant_b),min(plant_c))
print(lowest_moisture)
if lowest_moisture==min(plant_a):
    driest_plant="Plant A"
elif lowest_moisture==min(plant_b):
    driest_plant="Plant B"
else:
    driest_plant="Plant C"
print(driest_plant)
plt.text(1,95,f"Driest Plant: {driest_plant}")
plt.title("Smart Gardening Moisture Monitor")
plt.xlabel("Day")
plt.ylabel("Soil Moisture(%)")
plt.ylim(0.3,100)
plt.xlim(0,7)
plt.grid()
plt.legend()
plt.tight_layout()
plt.savefig("smart_garden_moisture.png", dpi=300, bbox_inches="tight")
plt.show()
```
